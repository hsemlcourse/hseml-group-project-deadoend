```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os

# 1. Загружаем данные
df = pd.read_csv('../data/raw/E_commerce.csv')

# 2. УМНАЯ ОЧИСТКА
# Список колонок, которые должны быть числами
numeric_cols = ['Sales', 'Profit', 'Shipping Cost', 'Quantity', 'Discount']

for col in numeric_cols:
    # Сначала убираем символы $ и запятые (используем r'' чтобы не было Warning)
    df[col] = df[col].astype(str).str.replace(r'[\$,]', '', regex=True)
    
    # Превращаем в числа. Если там текст типа '0.xf' или 'abc', он станет NaN (пустотой)
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Удаляем строки, где появились пустые значения (те самые '0.xf' и 'abc')
df = df.dropna(subset=numeric_cols)

# 3. Работа с датами
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
df = df.dropna(subset=['Order Date']) # удаляем если дата сломана
df['Month'] = df['Order Date'].dt.month

# 4. Признаки
df['Total_Engagement'] = df['Like'] + df['Share'] + df['Add to Cart']

# 5. График
plt.figure(figsize=(10, 6))
sns.histplot(df['Profit'], kde=True, color='green')
plt.title('Распределение прибыли (после очистки мусора)')
plt.show()

# 6. Сохраняем
os.makedirs('../data/processed', exist_ok=True)
df.to_csv('../data/processed/cleaned_data.csv', index=False)

print(f"Готово! Мусор удален. Осталось строк: {len(df)}")
```
